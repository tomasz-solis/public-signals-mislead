## How to read the result

The goal is not to claim a production-grade classifier.

The goal is to show that:

- search decay alone is weak
- combined public signals can add a little structure
- even then, the sample is small and true business value is often still unknown

That is why the repo is framed as decision support under incomplete information, not as a product-outcome prediction task.
